#Tensor Operation

#####**Basic operation**

```python
my_tensor = torch.tensor([1,2,3])
print(my_tensor)
print(my_tensor + 10)
```

```python
torch.add(my_tensor , 10)
```

```python
print(my_tensor - 4)
print(my_tensor *2)
print(torch.mul(my_tensor, 10))
print(my_tensor / 10)
```

####**Automatic differentiaion**

```python
x = torch.tensor(1.0, requires_grad=True)

def u(x):
  return x * x

def g(u):
  return -u
```

```python
dgdx = torch.autograd.grad(g(u(x)), x)[0]
print(dgdx)
```

####**matrix Multiplication**

```python
my_tensor = torch.tensor([1,2,3])

print(my_tensor * my_tensor)
print(torch.mul(my_tensor, my_tensor))
print(torch.matmul(my_tensor, my_tensor))
```

```python
tensor_a = torch.rand(4,8)
tensor_b = torch.rand(8,4)

print("shape a: ",tensor_a.shape, " shape b: ", tensor_b.shape)
print("--"*10)

matrix_multiplaction = torch.matmul(tensor_a, tensor_b)
print("shpae matrix_multiplaction: ", matrix_multiplaction.shape)

matrix_multiplaction = torch.matmul(tensor_b, tensor_a)
print("shpae matrix_multiplaction: ", matrix_multiplaction.shape)
```

```python
try :
  torch.matmul(tensor_a,tensor_a)
except Exception as err:
  print("Error: ", err)
print('input shape: ', tensor_a.shape)
```

```python
transpose_a = tensor_a.T
print(transpose_a.shape, tensor_a.shape)
torch.matmul(tensor_a, transpose_a)
```

```python
a = tensor_a.matmul(transpose_a)
b = torch.mm(tensor_a, transpose_a)
c = torch.matmul(tensor_a, transpose_a)

# a == b
a == c
```

#####**stacking**

```python
x = torch.arange(0, 6, 1)
x, x.shape
```

```python
stacked = torch.stack([x,x,x], dim=0)
print(stacked)
print()
print(stacked.shape)
```

```python
stacked = torch.stack([x,x,x], dim=1)
print(stacked)
print()
print(stacked.shape)
```

#####**Concatenation**

```python
x = torch.arange(0, 6, 1)
x, x.shape
```

```python
concat = torch.cat([x,x,x], dim=0)
print(concat)
print()
print(concat.shape)
```

```python
try:
  concat = torch.cat([x,x,x], dim=1)
except Exception as err:
  print("Error: ", err)
  print("shape: ", concat.shape)
```

```python
# use unsqueeze to slove problem in dim=1
# y = torch.unsqueeze(x, dim=1)
y = x.unsqueeze(dim=1)
print(y)
# print()
print(y.shape)
```

```python
concat = torch.cat([y,y,y], dim=1)
print(concat)
print()
print(concat.shape)
```
