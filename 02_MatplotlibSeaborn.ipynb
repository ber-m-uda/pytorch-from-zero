```python
import numpy as np
import matplotlib.pyplot as plt
```

#Line plot

```python
x = np.array([0,6,33])
y = np.array([10,5,45])

plt.plot(x,y)
plt.show()
```

```python
value = np.array([3, 8, 1, 10 ,5 ,15])

plt.plot(value, marker="*", linestyle='--', color='black')
plt.show()
```

```python
# sin
np.pi # = 3.14

x = np.arange(0, 5*np.pi, 0.1)
y = np.sin(x)

plt.plot(x,y)
plt.grid()
plt.show()
```

```python
# duble plot

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label='sine')
plt.plot(x, y_cos, label='cosine')

plt.xlabel('x')
plt.ylabel('y')

plt.title('sin and cos')
plt.legend()

# plt.legend(['sin','cos'])

plt.show()
```

```python
# sub plot

plt.subplot(2,2,1)
plt.plot(x,y_sin)
plt.title('sin')

plt.subplot(2,2,4)
plt.plot(x,y_cos)
plt.title('cos')

plt.tight_layout()
plt.show()
```

```python
# Color map - without aspect

gradient = np.linspace(0,1,256)
gradient = np.vstack((gradient,gradient))

plt.imshow(gradient, cmap='nipy_spectral')
plt.show()
```

```python
# colormap - with aspect

plt.imshow(gradient, aspect="auto", cmap='nipy_spectral')
plt.axis('off')
plt.title("colormap")
plt.show()
```

#Histogram

```python
x = np.random.normal(loc=100, scale=2, size=300)

plt.hist(x, color="teal", edgecolor="black")
plt.show()
```

#pie Chart

```python
y = np.array([35,25,25,15])
mylabels = ['Apples','Bananas','Cherries','Dates']

# plt.pie(y, labels=mylabels)
plt.pie(y, labels=None, colors=plt.cm.plasma_r(np.linspace(0,1,len(y))))
plt.legend(mylabels, loc=[1,0.5])

plt.show()
```
