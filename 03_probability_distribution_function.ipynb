```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

#PDF

```python
x = np.linspace(-4 ,4 ,1000)
mu = 0
sigma = 1
pdf_values = 1/ (np.sqrt(2*np.pi)*sigma) * np.exp(-0.5*((x-mu)/sigma)**2)

plt.figure(figsize=(8,6))
plt.plot(x, pdf_values, label='normal distribution')

plt.xlabel('x')
plt.ylabel('pdf(x)')
plt.title('normal distribution')

plt.grid(True)
plt.legend()
plt.show()
```

```python
data = np.random.normal(loc=5, scale=1, size=10_000_000) # samping from distribution

sns.displot(data, kind="kde")
plt.show()
```

```python
data = np.random.uniform(size=1_000)
sns.displot(data, kind="kde")
plt.show()
```
