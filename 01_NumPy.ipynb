#indexing and slicing

```python
# 1-D
a = np.array([1,2,3,4,5,6,7])
print("shape: ", a.shape)
print(a[0])
print(a[0:4])
print(a[:6])
print(a[:20])
print(a[2:5])
print(a[-3])
print(a[0:-3])
```

```python
# 2-D
a = np.array([[1,2,3,4],[5,6,7,8]])
print(a.shape)
print(a[0])
print(a[0,0], a[0,1], a[1,0])

a[0]= 6
print(a)

a[0,-1] = 77
print(a)
```
