joining array

````python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
print(arr1.shape)
print(arr2.shape)

arr = np.concatenate((arr1, arr2),axis=0)
# arr = np.stack((arr1, arr2))
print(arr)
````

````python
arr1 = np.array([[1,2],
                 [4,5]])

arr2 = np.array([[7,8],
                 [10,11]])
````

````python
arr = np.concatenate((arr1, arr2), axis=1)
print("array1 shape: ", arr1.shape)
print("array2 shape: ", arr2.shape)
print("result shape: ", arr.shape)
print(arr)
````

stack

````python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])


arr = np.stack((arr1, arr2), axis=1)
print("array1 shape: ", arr1.shape)
print("array2 shape: ", arr2.shape)
print("New shape: ", arr.shape)
print()
print(arr)
````
